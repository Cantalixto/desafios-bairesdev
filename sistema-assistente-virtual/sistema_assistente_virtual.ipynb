```python
def falar(texto):
    tts = gTTS(text=texto, lang='pt-br')
    arquivo = "resposta.mp3"
    tts.save(arquivo)
    return Audio(arquivo, autoplay=True)

# Função para fazer pergunta (texto do usuário)
def perguntar(pergunta):
    return falar(f"Você perguntou: {pergunta}")

# Função para comandos do assistente
def executar_comando(comando):
    comando = comando.lower()

    if "youtube" in comando:
        falar("Abrindo YouTube para você.")
        print("🔗 https://www.youtube.com")
        return falar("Aqui está o link do YouTube!")

    elif "hora" in comando or "horário" in comando:
        fuso_brasilia = pytz.timezone("America/Sao_Paulo")
        agora = datetime.datetime.now(fuso_brasilia)
        hora_formatada = agora.strftime("%H:%M")
        return falar(f"Agora são {hora_formatada} em Brasília.")

    else:
        return falar("Desculpe, não reconheci esse comando.")

```

```python
executar_comando("Abrir o YouTube")
```

```python
executar_comando("Que horas são?")
```

```python
import speech_recognition as sr

def ouvir_comando():
    recognizer = sr.Recognizer()
    with sr.Microphone() as source:
        print("🎙️ Ouvindo comando...")
        audio = recognizer.listen(source)

        try:
            comando = recognizer.recognize_google(audio, language='pt-BR')
            print(f"🗣️ Você disse: {comando}")
            return comando
        except sr.UnknownValueError:
            print("❌ Não entendi o que você disse.")
            return ""
        except sr.RequestError:
            print("⚠️ Erro ao acessar o serviço de reconhecimento.")
            return ""
```

```python
comando_voz = ouvir_comando()
if comando_voz:
    executar_comando(comando_voz)
```
